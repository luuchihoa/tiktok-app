```
#@title 🎬 3. Render Video MP4 (GPU T4 16GB Tăng Tốc)
import os
%cd /content/tiktok-app
!mkdir -p out

# Render tăng tốc GPU với OpenGL/EGL
!xvfb-run -a -s "-screen 0 1080x1920x24" npx remotion render CatholicVideo out/CatholicVideo.mp4 \n  --gl=angle \n  --concurrency=2 \n  --jpeg-quality=85 \n  --pixel-format=yuv420p \n  --video-bitrate=15M \n  --color-space=bt709 \n  --overwrite

print("🎉 Render Video Hoàn Tất 100%!")
```

```
#@title ⬇️ 4. Tự Động Tải Video MP4 Về Máy Tính
from google.colab import files
import os

video_file = "/content/tiktok-app/out/CatholicVideo.mp4"
if os.path.exists(video_file):
  size_mb = os.path.getsize(video_file) / (1024 * 1024)
  print(f"📦 Video sẵn sàng ({size_mb:.2f} MB). Đang kích hoạt tải về...")
  files.download(video_file)
else:
  print("❌ Không tìm thấy file video. Vui lòng kiểm tra lại bước render.")
```
