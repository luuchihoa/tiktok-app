```
#@title 🎬 3. Render Video MP4 (1080x1920 30FPS Tăng Tốc GPU)
%cd /content/tiktok-app
!mkdir -p out
!npx remotion render CatholicVideo out/CatholicVideo.mp4 \n  --gl=angle \n  --concurrency=4 \n  --video-bitrate=15M \n  --color-space=bt709 \n  --jpeg-quality=100 \n  --overwrite

print("🎉 Render Video Hoàn Tất 100%!")
```

```
#@title ⬇️ 4. Tự Động Tải Video MP4 Về Máy Tính
from google.colab import files
import os

video_file = "/content/tiktok-app/out/CatholicVideo.mp4"
if os.path.exists(video_file):
  size_mb = os.path.getsize(video_file) / (1024 * 1024)
  print(f"📦 Video sẵn sàng ({size_mb:.2f} MB). Đang kích hoạt tải về trình duyệt...")
  files.download(video_file)
else:
  print("❌ Không tìm thấy file video. Vui lòng kiểm tra lại bước render.")
```
